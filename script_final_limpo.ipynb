1. IMPORTAR PACOTES

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

from linearmodels.panel import PanelOLS, BetweenOLS, RandomEffects
from statsmodels.stats.diagnostic import het_breuschpagan
from scipy.stats import chi2
```

2. LIMPEZA E PREPARAÇÃO DA BASE

```python
# Carregar base
df = pd.read_csv("base_final.csv")

# Converter variáveis numéricas
for col in ["ideb", "fundeb", "total", "ano"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Criar FUNDEB por aluno
df["fundeb_por_aluno"] = df["fundeb"] / df["total"]
df["fundeb_por_aluno"] = df["fundeb_por_aluno"].replace([np.inf, -np.inf], np.nan)

# Log do FUNDEB por aluno
df["ln_fundeb_por_aluno"] = np.log(df["fundeb_por_aluno"])
df["ln_fundeb_por_aluno"] = df["ln_fundeb_por_aluno"].replace([np.inf, -np.inf], np.nan)

# Remover faltantes essenciais
df = df.dropna(subset=["ideb", "ln_fundeb_por_aluno", "codigo_municipio", "ano"]).copy()

print("Observações após limpeza:", df.shape[0])
df.head()
```

3. ESTATÍSTICAS DESCRITIVAS

```python
# contar municípios únicos

df["codigo_municipio"].nunique()
```

```python
variaveis_desc = ["ideb", "fundeb_por_aluno", "ln_fundeb_por_aluno"]

estatisticas_descritivas = df[variaveis_desc].describe().T
estatisticas_descritivas["Mediana"] = df[variaveis_desc].median()

estatisticas_descritivas = estatisticas_descritivas[
    ["mean", "std", "min", "Mediana", "max", "count"]
].round(4)

estatisticas_descritivas.columns = [
    "Média", "Desvio-Padrão", "Mínimo", "Mediana", "Máximo", "Observações"
]

print("===== ESTATÍSTICAS DESCRITIVAS =====")
print(estatisticas_descritivas)
```

4. TABELA DESCRITIVA

```python
estatisticas_descritivas.to_excel("tabela_descritiva.xlsx")
print("Arquivo exportado: tabela_descritiva.xlsx")
```

5. ESTIMAÇÃO DOS MODELOS

```python
# Estrutura de painel
df_panel = df.set_index(["codigo_municipio", "ano"]).sort_index()

# Variáveis
y = df_panel["ideb"].astype(float)
X = df_panel[["ln_fundeb_por_aluno"]].astype(float)
X_const = sm.add_constant(X)

# Modelos
ols = sm.OLS(y, X_const).fit(cov_type="HC1")
fe = PanelOLS(
    y, X,
    entity_effects=True,
    time_effects=True
).fit(cov_type="clustered", cluster_entity=True)

re = RandomEffects(y, X).fit(cov_type="robust")
between = BetweenOLS(y, X).fit(cov_type="robust")

print("Modelos estimados com sucesso.")
```

6. ESTRUTURAR TABELAS DOS MODELOS

```python
def estrelas_pvalor(p):
    if p < 0.01:
        return "***"
    elif p < 0.05:
        return "**"
    elif p < 0.10:
        return "*"
    return ""

def coef_ep(coef, ep, p):
    return f"{coef:.4f}{estrelas_pvalor(p)}\n({ep:.4f})"

tabela_modelos = pd.DataFrame({
    "OLS": [
        coef_ep(ols.params["const"], ols.bse["const"], ols.pvalues["const"]),
        coef_ep(ols.params["ln_fundeb_por_aluno"], ols.bse["ln_fundeb_por_aluno"], ols.pvalues["ln_fundeb_por_aluno"]),
        "Robusto (HC1)",
        f"{ols.rsquared:.4f}",
        f"{int(ols.nobs)}"
    ],
    "FE (two-way)": [
        "-",
        coef_ep(fe.params["ln_fundeb_por_aluno"], fe.std_errors["ln_fundeb_por_aluno"], fe.pvalues["ln_fundeb_por_aluno"]),
        "Cluster (município)",
        f"{fe.rsquared:.4f}",
        f"{int(fe.nobs)}"
    ],
    "RE": [
        "-",
        coef_ep(re.params["ln_fundeb_por_aluno"], re.std_errors["ln_fundeb_por_aluno"], re.pvalues["ln_fundeb_por_aluno"]),
        "Robusto",
        f"{re.rsquared:.4f}",
        f"{int(re.nobs)}"
    ],
    "Between": [
        "-",
        coef_ep(between.params["ln_fundeb_por_aluno"], between.std_errors["ln_fundeb_por_aluno"], between.pvalues["ln_fundeb_por_aluno"]),
        "Robusto",
        f"{between.rsquared:.4f}",
        f"{int(between.nobs)}"
    ]
}, index=[
    "Constante",
    "ln(FUNDEB por aluno)",
    "Tipo de erro-padrão",
    "R²",
    "Observações"
])

print("===== TABELA DOS MODELOS =====")
print(tabela_modelos)
```

7. TESTE ECONOMÉTRICOS

```python
# Teste F do modelo FE
f_stat = fe.f_statistic.stat if fe.f_statistic is not None else np.nan
f_pval = fe.f_statistic.pval if fe.f_statistic is not None else np.nan

# Teste de Hausman
b = fe.params
B = re.params

coef_comuns = b.index.intersection(B.index)
b = b[coef_comuns]
B = B[coef_comuns]

V_b = fe.cov.loc[coef_comuns, coef_comuns]
V_B = re.cov.loc[coef_comuns, coef_comuns]

diff = b - B
V_diff = V_b - V_B

try:
    hausman_stat = float(diff.T @ np.linalg.inv(V_diff) @ diff)
    hausman_pval = 1 - chi2.cdf(hausman_stat, len(diff))
except np.linalg.LinAlgError:
    hausman_stat = np.nan
    hausman_pval = np.nan

# Teste de Breusch-Pagan
bp_lm_stat, bp_lm_pval, bp_f_stat, bp_f_pval = het_breuschpagan(ols.resid, X_const)

print("Testes calculados com sucesso.")
```

8. ESTRUTURANDO A TABELA DOS TESTES

```python
tabela_testes = pd.DataFrame({
    "Estatística": [f_stat, hausman_stat, bp_lm_stat, bp_f_stat],
    "p-valor": [f_pval, hausman_pval, bp_lm_pval, bp_f_pval]
}, index=[
    "F (efeitos fixos)",
    "Hausman (FE vs RE)",
    "Breusch-Pagan (LM)",
    "Breusch-Pagan (F)"
]).round(4)

print("===== TABELA DOS TESTES =====")
print(tabela_testes)
```
